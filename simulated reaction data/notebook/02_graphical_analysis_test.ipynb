```python
import matplotlib.pyplot as plt
import pandas as pd

data = pd.read_csv('reaction_data_copy.csv')

plt.scatter(data['Temperature (K)'], data['Conversion A to B'])
plt.xlabel('Temperature (K)')
plt.ylabel('Conversion A to B')
plt.title('Conversion A to B vs Temperature')
plt.show()
```

```python
plt.scatter(data['Reaction Rate'], data['Conversion A to B'])
plt.xlabel('Reaction Rate')
plt.ylabel('Conversion A to B')
plt.title('Conversion A to B vs Reaction Rate')
plt.show()
```
