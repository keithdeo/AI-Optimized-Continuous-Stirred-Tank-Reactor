```python
def heating_energy(mass, cp, feed_temperature, reactor_temperature):
    energy = mass * cp * (
        reactor_temperature - feed_temperature
    )

    return energy
```

```python
# all example values
mass = 10
cp = 4.18
feed_temperature = 300
reactor_temperature = 500

energy = heating_energy(
    mass,
    cp,
    feed_temperature,
    reactor_temperature
)

print("Heating energy required:", energy, "kJ")
```
